**1. 讀取有效書籍**

```python
from pymarc import MARCReader

books_valid = []
with open('../data/raw/books.mrc', 'rb') as fh:
    reader = MARCReader(fh, to_unicode=True, force_utf8=True)
    for record in reader:
        if record['001'] is None:
            continue
        if record.title is None:
            continue
        books_valid.append({
            "id": record['001'].value(),
            "contents": record.title
        })
print(f"共有 {len(books_valid)} 本有效書籍")
```

**2. 移除有重複書名的書籍**

```python
books = []
for book in books_valid:
    if book['contents'] not in [b['contents'] for b in books]:
        books.append(book)
print(f"去除重複書名後，剩餘 {len(books)} 本書籍")
```

**3. 將處理好的書籍轉成 json 並儲存**

```python
import json

books_json = json.dumps(books, ensure_ascii=False, indent=4)
with open('../data/processed/books.json', 'w', encoding='utf-8') as f:
    f.write(books_json)
print(f"已將書籍資料寫入 ../data/processed/books.json")
```
