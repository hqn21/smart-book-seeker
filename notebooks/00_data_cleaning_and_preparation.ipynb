**1. Load books**

```python
from pymarc import MARCReader

books_valid = []
with open('../data/raw/books.mrc', 'rb') as fh:
    reader = MARCReader(fh, to_unicode=True, force_utf8=True)
    for record in reader:
        if record['001'] is None:
            continue
        if record.title is None:
            continue
        books_valid.append({
            "id": record['001'].value(),
            "title": record.title
        })
print(f"There are {len(books_valid)} available books loaded")
```

**2. Remove duplicate books**

```python
books = []
for book in books_valid:
    if book['title'] not in [b['title'] for b in books]:
        books.append(book)
print(f"After removing duplicate books, there are {len(books)} books left")
```

**3. Convert processed books into json file**

```python
import json

metadata = {
    "index": {
        "_index": "books",
        "_id": ""
    }
}
with open('../data/processed/books.jsonl', 'w', encoding='utf-8') as f:
    for book in books:
        metadata["index"]["_id"] = book['id']
        f.write(json.dumps(metadata, ensure_ascii=False))
        f.write('\n')
        f.write(json.dumps(book, ensure_ascii=False))
        f.write('\n')
print(f"Processed books are saved at ../data/processed/books.jsonl")
```
